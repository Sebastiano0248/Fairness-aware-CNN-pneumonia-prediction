```python
import os
import pandas as pd
```

```python
df = pd.read_csv('../../data/CheXpert-v1.0-small_raw/train.csv')

# Frontal views only (AP and PA)
df = df[df['Frontal/Lateral'] == 'Frontal']

# Extract patient_id from path
df['patient_id'] = df['Path'].str.extract(r'(patient\d+)')

# Extract study_id from path
df['study_id'] = df['Path'].str.extract(r'(study\d+)')

# Strip 'CheXpert-v1.0-small/' prefix to get path relative to data/CheXpert-v1.0-small_raw/
df['image'] = df['Path'].str.replace('CheXpert-v1.0-small/', '', regex=False)

df.head()
```

```python
# Positives: Pneumonia == 1.0 (exclude uncertain -1.0 and unmentioned NaN)
# One image per patient: keep the first study alphabetically
df_pos = df[df['Pneumonia'] == 1.0]
df_pos = df_pos.loc[df_pos.groupby('patient_id')['study_id'].idxmin()]

display(df_pos[['image', 'Sex', 'Age', 'AP/PA', 'Pneumonia']].describe())
df_pos
```

```python
# Negatives: No Finding == 1.0, patients with a single study only
# (mirrors NIH logic of single follow-up patients)
study_counts = df.groupby('patient_id')['study_id'].nunique().reset_index()
study_counts.rename(columns={'study_id': 'n_studies'}, inplace=True)
single_study_patients = study_counts[study_counts['n_studies'] == 1]['patient_id']

df_neg = df[
    (df['patient_id'].isin(single_study_patients)) &
    (df['No Finding'] == 1.0)
]
# One image per patient
df_neg = df_neg.loc[df_neg.groupby('patient_id')['study_id'].idxmin()]

display(df_neg[['image', 'Sex', 'Age', 'AP/PA', 'No Finding']].describe())
df_neg
```

```python
df_neg = df_neg.sample(frac=0.1, random_state=42)
```

```python
df_out = pd.concat([df_pos, df_neg])
df_out['class'] = df_out['Pneumonia'].apply(lambda x: 1 if x == 1.0 else 0)

# Drop rows with missing or unknown gender
df_out = df_out[df_out['Sex'].isin(['Male', 'Female'])]
df_out['gender'] = df_out['Sex'].map({'Male': 0, 'Female': 1})

display(df_out.describe())
df_out
```

```python
df_out = df_out.sample(frac=1, random_state=42).reset_index(drop=True)
df_out = df_out.rename(columns={
    'Age': 'age',
    'AP/PA': 'view_pos'
})
df_out = df_out[['image', 'age', 'gender', 'view_pos', 'class']]
df_out
```

```python
df_out.to_csv('../../data/CheXpert-v1.0-small_raw/experiment_data.csv', index=False)
```
